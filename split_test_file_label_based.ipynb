```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path(r"C:\Users\Bruker\msc_data")

paths_dict = {
            "left_arm": BASE_DIR / "prelim_7" / "Muse_E2511_RED-ARM_LEFT_new_time_hz_labeled_median_filter_no_idle_cleaned.csv",
            "right_arm": BASE_DIR / "prelim_7" / "Muse_E2511_GREY-ARM_RIGHT_new_time_hz_labeled_median_filter_no_idle_cleaned.csv",
            "upper_back": BASE_DIR / "prelim_7" / "muse_v3-UB_new_time_hz_labeled_median_filter_no_idle_rotated_cleaned.csv",
            "lower_back": BASE_DIR / "prelim_7" / "muse_v3_3-LB_new_time_hz_labeled_median_filter_no_idle_rotated_cleaned.csv",
            "left_fsr": BASE_DIR / "prelim_7" / "mitch_B0308-LEFT_del_end_dupli_new_time_labeled_median_filter_no_idle_cleaned.csv",
            "right_fsr": BASE_DIR / "prelim_7" / "mitch_B0510-RIGHT_del_end_dupli_new_time_labeled_median_filter_no_idle_cleaned.csv"
        }

prelim_labels = ["forward_lean", "sideways_lean", "torso_twist", "backward_lean", "arms_forward", "shoulder_load", "drag", "neutral_load_left",
                 "neutral_load_right", "stairs_up", "stairs_down"]

larm_df = pd.read_csv(paths_dict["left_arm"])
rarm_df = pd.read_csv(paths_dict["right_arm"])
lb_df = pd.read_csv(paths_dict["lower_back"])
ub_df = pd.read_csv(paths_dict["upper_back"])
lsole_df = pd.read_csv(paths_dict["left_fsr"])
rsole_df = pd.read_csv(paths_dict["right_fsr"])
```

```python
# filter dfs
dfs_and_paths = {
    "left_arm": (larm_df, paths_dict["left_arm"]),
    "right_arm": (rarm_df, paths_dict["right_arm"]),
    "lower_back": (lb_df, paths_dict["lower_back"]),
    "upper_back": (ub_df, paths_dict["upper_back"]),
    "left_fsr": (lsole_df, paths_dict["left_fsr"]),
    "right_fsr": (rsole_df, paths_dict["right_fsr"]),
}

for sensor_name, (df, original_path) in dfs_and_paths.items():

    # filter
    filtered_df = df[df["label"].isin(prelim_labels)].copy()

    # construct new filename
    original_path = Path(original_path)

    new_path = (
        original_path.parent /
        f"{original_path.stem}_prelim_only.csv"
    )

    # save
    filtered_df.to_csv(new_path, index=False)

    print(f"Saved {sensor_name} -> {new_path}")
```
